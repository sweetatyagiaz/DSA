#### Queue Implementation

```python
class Queue():
    def __init__(self, k):
        self.k = k
        self.queue = [None]*k
        self.head = self.tail = -1

    # Insert an element into queue
    def enqueue(self, data):
        if (self.tail == self.k - 1):
            print('The Queue is full\n')

        elif self.head == -1:
            self.head = 0
            self.tail = 0
            self.queue[self.tail] = data
        else:
            self.tail = self.tail + 1
            self.queue[self.tail] = data


    # Delete an element
    def dequeue(self):
        if self.head == -1:
            print("The Queue is empty\n")

        elif self.head == self.tail:
            temp = self.queue[self.head]
            self.head = -1
            self.tail = -1
            return temp
        else:
            temp = self.queue[self.head]
            self.head = self.head + 1
            return temp

    # Print Queue
    def printQueue(self):
        if self.head == -1:
            print("Queue is empty")
        else:
            for idx in range(self.head, self.tail + 1):
                print(self.queue[idx], end=" ")
            print()
```

```python
# Your Queue object will be instantiated and called as such:
obj = Queue(5)
obj.enqueue(1)
obj.enqueue(2)
obj.enqueue(3)
obj.enqueue(4)
obj.enqueue(5)
print("Initial queue")
obj.printQueue()

obj.dequeue()
print("After removing an element from the queue")
obj.printQueue()
```
