### **Heap Data Structure**

Heap Insert Algorithm (Max Heap)
1. Insert the new element at the end of the heap.
2. Compare it with its parent.
3. If the new element is greater than the parent, swap them.
4. Continue until the heap property is restored or the root is reached.

| Operation   | Complexity |
| ----------- | ---------- |
| Insert      | O(log n)   |
| Access Root | O(1)       |
| Delete Root | O(log n)   |


```python
class MaxHeap:
    def __init__(self):
        self.heap = []

    def insert_max(self, value):
        # Insert at the end
        self.heap.append(value)

        # Heapify Up
        index = len(self.heap) - 1

        while index > 0:
            parent = (index - 1) // 2

            if self.heap[index] > self.heap[parent]:
                self.heap[index], self.heap[parent] = (
                    self.heap[parent],
                    self.heap[index]
                )
                index = parent
            else:
                break

    def insert_min(self, value):
            # Insert at the end
            self.heap.append(value)
    
            # Heapify Up
            index = len(self.heap) - 1
    
            while index > 0:
                parent = (index - 1) // 2
    
                if self.heap[index] < self.heap[parent]:
                    self.heap[index], self.heap[parent] = (
                        self.heap[parent],
                        self.heap[index]
                    )
                    index = parent
                else:
                    break

    def display(self):
        print(self.heap)
```

```python
# Example: Max Heap
heap = MaxHeap()

heap.insert_max(10)
heap.insert_max(20)
heap.insert_max(15)
heap.insert_max(30)

heap.display()
```

```python
# Example: Min Heap
heap = MaxHeap()

heap.insert_min(10)
heap.insert_min(20)
heap.insert_min(15)
heap.insert_min(30)

heap.display()
```
